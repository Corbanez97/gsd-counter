```python
import torch
import numpy as np
```

```python
shape = (4,4)
rand_tensor = np.round(torch.rand(shape))

print(f"Random Tensor: \n {rand_tensor} \n")

print(f"Roll +e_1 \n{torch.roll(rand_tensor, 1, 1)}\n")
print(f"Roll -e_1 \n{torch.roll(rand_tensor, -1, 1)}\n")  
print(f"Roll -e_2 \n{torch.roll(rand_tensor, 1, 0)}\n")
print(f"Roll  +e_2 \n{torch.roll(rand_tensor, -1, 0)}\n") 
print(f"Roll  +e_2+e_1 \n{torch.roll(torch.roll(rand_tensor, -1, 0), 1, 1)}\n") 
print(f"Roll  -e_2-e_1 \n{torch.roll(torch.roll(rand_tensor, 1, 0), -1, 1)}\n") 

print(f"Overall sum for all six neighbooring sites \n{torch.remainder(torch.roll(rand_tensor, 1, 1) + torch.roll(rand_tensor, -1, 1)+ torch.roll(rand_tensor, 1, 0) + torch.roll(rand_tensor, -1, 0) + torch.roll(torch.roll(rand_tensor, -1, 0), 1, 1) + torch.roll(torch.roll(rand_tensor, 1, 0), -1, 1), 2)}")
```
